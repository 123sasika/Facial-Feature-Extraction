```python
from mtcnn import MTCNN
```

```python
image_path = "ip.jpg"
image = cv2.imread(image_path)
```

```python
image
```

```python
output_dir = "face_parts"
os.makedirs(output_dir, exist_ok=True)
```

```python
detector = MTCNN()
```

```python
results = detector.detect_faces(image)
```

```python
results
```

```python
for i, result in enumerate(results):
    keypoints = result['keypoints']
print(result)
print(keypoints)
```

```python
for part_name, (x, y) in keypoints.items():
    size = 40
    x1, y1 = max(x - size, 0), max(y - size, 0)
    x2, y2 = min(x + size, image.shape[1]), min(y + size, image.shape[0])
    
    part_img = image[y1:y2, x1:x2]
    
    part_filename = os.path.join(output_dir, f"face_{i}_{part_name}.jpg")
    cv2.imwrite(part_filename, part_img)
```

```python
print("Facial parts have been extracted and saved.")
```

```python
for part_name, (x, y) in keypoints.items():
    size = 40
    x1, y1 = max(x - size, 0), max(y - size, 0)
    x2, y2 = min(x + size, image.shape[1]), min(y + size, image.shape[0])

    cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)

    cv2.putText(image, part_name, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

cv2.imwrite("annotated_face.jpg", image)
```
